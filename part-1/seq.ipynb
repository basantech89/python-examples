```python
from functools import lru_cache


@lru_cache(2**10)
def fib(n: int) -> int:
    return 1 if n < 2 else fib(n - 1) + fib(n - 2)


fib(100)
```

```python
class Fib:
    def __init__(self, n: int) -> None:
        self.n = n

    def __len__(self) -> int:
        return self.n

    def __getitem__(self, s: int | slice) -> int | list[int]:
        if isinstance(s, int):
            if s < 0:
                s = self.n + s

            if s < 0 or s >= self.n:
                raise IndexError("Index out of range")

            return Fib._fib(s)

        start, stop, step = s.indices(self.n)
        print(f"Slice indices: start={start}, stop={stop}, step={step}")
        rng = range(start, stop, step)
        # rng = range(*s.indices(self.n))
        return [Fib._fib(i) for i in rng]

    @staticmethod
    @lru_cache(2**10)
    def _fib(n: int) -> int:
        return 1 if n < 2 else Fib._fib(n - 1) + Fib._fib(n - 2)
```

```python
from typing import Any


class MyClass:
    def __init__(self, name: str) -> None:
        self.name = name

    def __repr__(self) -> str:
        return f"MyClass(name={self.name!r})"

    def __add__(self, other: MyClass):
        return MyClass(self.name + other.name)

    def __iadd__(self, other: MyClass | str):
        if isinstance(other, MyClass):
            self.name += other.name
        else:
            self.name += other

        return self

    def __mul__(self, n: int):
        return MyClass(self.name * n)

    def __imul__(self, n: int):
        self.name *= n
        return self

    def __rmul__(self, n: int):
        return self.__mul__(n)

    def __contains__(self, value: Any):
        # if the type of value is not compatible with the type of elements in self.name,
        # I'll let Python handle it and it will raise an error
        return value in self.name


c1 = MyClass("Eric")
c2 = MyClass("Idle")

id(c1), id(c2)

result = c1 + c2
id(c1), id(c2)
result, id(result)

c1 += c2
id(c1), id(c2)
c1
c2

"----------- mul ------------"

d1 = MyClass("Eric")
d2 = MyClass("Idle")

id(d1), id(d2)
result = d1 * 3
result, id(result)

d2 *= 3
d2, id(d2)

e1 = MyClass("Eric")
# wouldn't be possible without the __rmul__ method
e1 * 3
3 * e1

"-------------- in ------------"
f1 = MyClass("Eric Idle")
"Eric" in f1
"X" in f1
```

```python
import numbers
from typing import Any


class Point:
    def __init__(self, x: Any, y: Any) -> None:
        if isinstance(x, numbers.Real) and isinstance(y, numbers.Real):
            self._pt = (x, y)
        else:
            raise TypeError("Point coordinates x and y must be real numbers")

    def __repr__(self) -> str:
        return f"Point(x={self._pt[0]}, y={self._pt[1]})"

    def __len__(self) -> int:
        return len(self._pt)

    def __getitem__(self, key: int | slice):
        return self._pt[key]


pt1 = Point(10, 20)
# to unpack pt1 into x, and y; Point must be an iterable
x, y = pt1
x, y
# since now Point implements __len__ and __getitem__, it can be unpacked like a sequence
# and now I can unpack Point like this as well
pt2 = Point(*pt1)
pt2
```

```python
from collections.abc import Iterable
from typing import Self

type IterablePoint = tuple[float, float] | list[float] | Point


class Polygon:
    def __init__(self, *pts: IterablePoint) -> None:
        self._pts = [Point(*pt) for pt in pts] if pts else []

    def __repr__(self) -> str:
        pts_str = ", ".join(map(str, self._pts))
        return f"Polygon({pts_str})"

    def __len__(self):
        return len(self._pts)

    def __getitem__(self, key: int | slice):
        return self._pts[key]

    def __setitem__(
        self, key: int | slice, value: Self | Point | Iterable[IterablePoint]
    ):
        try:
            rhs = [Point(*pt) for pt in value]
            is_single = False
        except TypeError:
            try:
                rhs = Point(*value)
                is_single = True
            except TypeError:
                raise TypeError("Invalid Point or iterable of Points")

        if (isinstance(key, int) and is_single) or (
            isinstance(key, slice) and not is_single
        ):
            self._pts = rhs
        else:
            raise TypeError("Incompatible index/slice assignment")

    def __add__(self, other: Any):
        if isinstance(other, Polygon):
            return Polygon(*self._pts, *other._pts)
        else:
            raise TypeError("Can only concatenate with other Polygon")

    def __iadd__(self, pt: Self | Iterable[IterablePoint]):
        self.extend(pt)
        return self

    def __delitem__(self, key: int | slice):
        del self._pts[key]

    def pop(self, key: int):
        return self._pts.pop(key)

    def append(self, pt: IterablePoint):
        self._pts.append(Point(*pt))

    def insert(self, index: int, pt: IterablePoint):
        self._pts.insert(index, Point(*pt))

    def extend(self, pts: Self | Iterable[IterablePoint]):
        # in case if pts is not an instance of Polygon, and
        # if pt doesn't has exactly 2 floats in it, then
        # I'll let Python handle it and it will raise an error
        points = pts._pts if isinstance(pts, Polygon) else [Point(*pt) for pt in pts]
        self._pts += points


p1 = Polygon((0, 0), Point(1, 1), [2, 2])
p1
p1[1]
p1[0:2]
p1[::-1]

p2 = Polygon(Point(3, 3))
id(p1), id(p2)
result = p1 + p2
id(p1), id(p2), id(result)

"------------ in place concatenation -------------"

p1 = Polygon((0, 0), Point(1, 1))
p2 = Polygon([2, 2], Point(3, 3))
id(p1), id(p2)
p1 += p2
p1, p2
id(p1), id(p2)

"-------------- in place concatenation with iterables ------------------"

p1 = Polygon((0, 0), Point(1, 1))
id(p1)
p1 += [(2, 2), [3, 3], Point(4, 4)]
p1, id(p1)

"---------------- append, extend, and insert ------------------"

p1 = Polygon((0, 0), Point(1, 1))
p2 = Polygon([2, 2], Point(3, 3))
id(p1)
p1.append([10, 10])
p1, id(p1)
p1.insert(1, Point(-1, -1))
p1, id(p1)
p1.extend(p2)
p1, p2
id(p1), id(p2)
p1.extend([(0, 0), Point(20, 20)])
id(p1), p1

"----------------- set item -------------"
p = Polygon((0, 0), (1, 1), (2, 2))
id(p), p
p[0:2] = [(10, 10), Point(20, 20), [30, 30]]
id(p), p
p[0] = Point(0, 0)
id(p), p
p[0] = Point(-1, -1)
id(p), p

"-------------- set item error handling -------------"

p = Polygon((0, 0), (1, 1), (2, 2))
# p[0] = [(0, 0), (1, 1)]
# p[0:2] = Point(0, 0)
# p[0] = ("a", "b")

"------------ delete item -------------"

p = Polygon((0, 0), (1, 1), (2, 2), (3, 3))
del p[0]
p
del p[0:2]
p

p.pop(0)
p
```

```python
class MyClass:
    def __init__(self, name: str, val: int) -> None:
        self.name = name
        self.val = val

    def __repr__(self) -> str:
        return f"MyClass({self.name}, {self.val})"

    def __lt__(self, other: MyClass):
        return self.val < other.val


c1 = MyClass("c1", 20)
c2 = MyClass("c2", 10)
c3 = MyClass("c3", 20)
c4 = MyClass("c4", 10)

c1 < c2
c1 > c2
sorted([c1, c4, c3, c2])
```
