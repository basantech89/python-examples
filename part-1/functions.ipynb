```python
l = [5, 8, 6, 10, 9]

_max: Callable[[int, int], int] = lambda x, y: max(y, x)


def max_sequence(sequence: list[int]) -> int:
    result = sequence[0]
    for item in sequence[1:]:
        result = _max(result, item)

    return result


max_sequence(l)
```

```python
def _reduce(fn: Callable[[int, int], int], sequence: list[int]) -> int:
    result = sequence[0]
    for item in sequence[1:]:
        result = fn(result, item)

    return result


_reduce(max, l)
```
