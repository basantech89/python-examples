```python
from collections import namedtuple

Color = namedtuple("Color", "red green blue alpha")


def random_color():
    red = randint(0, 255)
    green = randint(0, 255)
    blue = randint(0, 255)
    alpha = round(random(), 2)

    return Color(red, green, blue, alpha)


color = random_color()
color
```

```python
from typing import Any

data_list = [
    {"key2": 1, "key1": 2},
    {"key2": 4, "key1": 5},
    {"key3": 7, "key2": 8, "key1": 9},
    {"key2": 10},
]

keys = set(key for dicts in data_list for key in dicts.keys())

Struct = namedtuple("Struct", sorted(keys))
Struct._fields

Struct.__new__.__defaults__ = (None,) * len(Struct._fields)
Struct(key3=10)

tuple_list = [Struct(**d) for d in data_list]
tuple_list


def tuplify_dicts(dicts: dict[str, Any]):
    keys = set(key for dict_ in dicts for key in dict_.keys())
    Struct = namedtuple("Struct", sorted(keys), rename=True)

    Struct.__new__.__defaults__ = (None,) * len(Struct._fields)
    return [Struct(**dict_) for dict_ in data_list]


tuplify_dicts(data_list)
```
