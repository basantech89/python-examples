```python
import math

class Polygon:
  def __init__(self, vertices: int, radius: int):
    if vertices < 3:
      raise ValueError("Polygon must have at least 3 vertices.")

    self._vertices = vertices
    self._radius = radius

  def __repr__(self):
    return f"Polygon(vertices={self._vertices}, radius={self._radius})"

  @property
  def vertices(self):
    return self._vertices

  @property
  def radius(self):
    return self._radius

  @property
  def interior_angle(self):
    return (self._vertices - 2) * 180 / self._vertices

  @property
  def side_length(self):
    return 2 * self._radius * math.sin(math.pi / self._vertices)
  
  @property
  def apothem(self):
    return self._radius * math.cos(math.pi / self._vertices)

  @property
  def area(self):
    return self._vertices * self.side_length * self.apothem / 2

  @property
  def perimeter(self):
    return self._vertices * self.side_length

  def __eq__(self, value: Polygon):
    if isinstance(value, self.__class__):
      return self._vertices == value._vertices and self._radius == value._radius

    return NotImplemented

  def __gt__(self, value: Polygon):
    if isinstance(value, self.__class__):
      return self._vertices > value._vertices

    return NotImplemented

def test_polygon():
  rel_tol = 0.001
  abs_tol = 0.001

  vertices = 3
  radius = 1
  p = Polygon(vertices, radius)

  assert str(p) == f"Polygon(vertices={vertices}, radius={radius})", f"actual: {str(p)}"
  assert p.vertices == vertices, (f"actual: {p.vertices}, "
                                  f"expected: {vertices}")
  assert p.radius == radius
  assert p.interior_angle == 60

  vertices = 4
  radius = 1
  p = Polygon(vertices, radius)

  assert math.isclose(p.area, 2.0, rel_tol=rel_tol, abs_tol=abs_tol), f"actual: {p.area}, expected: {2.0}"

  p1 = Polygon(3, 10)
  p2 = Polygon(10, 10)
  p3 = Polygon(15, 10)
  p4 = Polygon(15, 100)
  p5 = Polygon(15, 100)

  assert p2 > p1
  assert p2 < p3
  assert p3 != p4
  assert p4 == p5

  try:
    p = Polygon(2, 10)
    assert False, "Polygon with 2 vertices should raise ValueError"
  except ValueError:
    pass
  
test_polygon()
```

```python
class Polygons:
  def __init__(self, max_vertices: int, radius: int):
    self._largest_polygon_vertices = max_vertices
    self._radius = radius
    self._polygons = [Polygon(vertices, radius) for vertices in range(3, max_vertices + 1)]

  @property
  def max_efficiency_polygon(self):
    return max(self._polygons, key=lambda polygon: polygon.area / polygon.perimeter)

  def __repr__(self):
    return f"Polygons(max_vertices={self._largest_polygon_vertices}, radius={self._radius})"

  def __len__(self):
    return self._largest_polygon_vertices - 2

  def __getitem__(self, key: int | slice):
    return self._polygons[key]

p = Polygons(5, 2)
p.max_efficiency_polygon
```
