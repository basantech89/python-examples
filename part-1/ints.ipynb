```python
digit_map = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz"


def encode(digits: list[int], digit_map: str = digit_map) -> str:
    if max(digits) >= len(digit_map):
        raise ValueError("Digit value exceeds digit map length")

    return "".join(digit_map[digit] for digit in digits)


def transform_to_base(n: int, base: int) -> str:
    if base < 2 or base > len(digit_map):
        raise ValueError(f"base must be between 2 and {len(digit_map)}")

    transformed: list[int] = []

    sign = -1 if n < 0 else 1
    numerator = n * sign
    modulus = None

    while numerator > 0:
        numerator, modulus = divmod(numerator, base)
        transformed.insert(0, modulus)

    encoding = encode(transformed, digit_map)
    return ("-" if sign == -1 else "") + encoding
```

```python
import time
from decimal import Decimal


def run_float(n=1):
    for i in range(n):
        a = 3.1415


def run_decimal(n=1):
    for i in range(n):
        a = Decimal("3.1415")


n = 10000000

start = time.perf_counter()
run_float(n)
end = time.perf_counter()
print(f"float operation time for {n} iterations: {end - start:.6f} seconds")

start = time.perf_counter()
run_decimal(n)
end = time.perf_counter()
print(f"decimal operation time for {n} iterations: {end - start:.6f} seconds")
```

```python
import math
import time
from decimal import Decimal


def run_float(n=1):
    a = 3.1415
    for i in range(n):
        math.sqrt(a)


def run_decimal(n=1):
    a = Decimal("3.1415")
    for i in range(n):
        a.sqrt()


n = 5000000

start = time.perf_counter()
run_float(n)
end = time.perf_counter()
print(f"float operation time for {n} iterations: {end - start:.6f} seconds")

start = time.perf_counter()
run_decimal(n)
end = time.perf_counter()
print(f"decimal operation time for {n} iterations: {end - start:.6f} seconds")
```
