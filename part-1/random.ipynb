```python
def generate_random_stuff(seed: int | None = None):
    random.seed(seed)
    result = []

    for _ in range(5):
        result.append(random.randint(0, 5))

    characters = list("abc")
    random.shuffle(characters)
    result.append(characters)

    for _ in range(5):
        result.append(random.gauss(0, 1))

    return result


generate_random_stuff()
generate_random_stuff()
```

```python
generate_random_stuff(0)
generate_random_stuff(0)
```

```python
from typing import Any


def freq_analysis(lst: list[Any]):
    return {k: lst.count(k) for k in set(lst)}


lst = [random.randint(0, 10) for _ in range(100)]
freq_analysis(lst)

lst = [random.randint(0, 10) for _ in range(1_000_000)]
d = freq_analysis(lst)
d

total = sum(d.values())
{k: v / total * 100 for k, v in d.items()}
```

```python
from typing import Any, NamedTuple

l = ["a", "b", "c"]


class Freq(NamedTuple):
    counter: int
    freq: float


def freq_counts(lst: list[Any]):
    total = len(lst)
    return {k: Freq(lst.count(k), lst.count(k) / total * 100) for k in set(lst)}


weights = [8, 1, 1]
cum_weights = [7, 8, 10]

freq_counts(random.choices(l, k=100_000))
freq_counts(random.choices(l, k=100_000, weights=weights))
freq_counts(random.choices(l, k=100_000, cum_weights=cum_weights))
```

```python
import random
from time import perf_counter

random.seed(0)

denominators = random.choices([0, 1], k=10_000_000, weights=[9, 1])
```

```python
start = perf_counter()
result = 0

for d in denominators:
    if d == 0:
        continue
    else:
        result = 10 / d

end = perf_counter()

print(f"Average elapsed time: {(end - start) / len(denominators):0.15f}")
```

```python
start = perf_counter()
result = 0

for d in denominators:
    try:
        result = 10 / d
    except ZeroDivisionError:
        pass

end = perf_counter()

print(f"Average elapsed time: {(end - start) / len(denominators):0.15f}")
```
