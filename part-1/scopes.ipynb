```python
from collections.abc import Callable
from functools import wraps


def counter(fn: Callable[..., object]):
    count = 0

    @wraps(fn)
    def inner(*args: Any, **kwargs: Any):
        nonlocal count

        count += 1
        print(f"Function {fn.__name__} called {count + 1} times")

        return fn(*args, **kwargs)

    return inner


@counter
def add(a: int, b: int) -> int:
    """Adds two numbers together.

    Args:
        a (int): The first number.
        b (int): The second number.

    Returns:
        int: The sum of the two numbers."""
    return a + b


@counter
def mul(a: int, b: int) -> int:
    """Multiplies two numbers together.

    Args:
        a (int): The first number.
        b (int): The second number.

    Returns:
        int: The product of the two numbers."""
    return a * b


help(add)
add(10, 20)
```

```python
from collections.abc import Callable
from functools import reduce


def timer(fn: Callable[..., int]):

    @wraps(fn)
    def inner(*args: int, **kwargs: int):
        start = perf_counter()
        result = fn(*args, **kwargs)
        end = perf_counter()

        args_ = ", ".join(map(str, args))
        kwargs_ = ",".join(f"{k}={v}" for k, v in kwargs.items())
        all_args = ", ".join(filter(None, [args_, kwargs_]))

        print(f"{fn.__name__}({all_args}) took {end - start:.6f} seconds")
        return result

    return inner


def calc_fib_recursive(n: int) -> int:
    return 1 if n <= 2 else calc_fib_recursive(n - 1) + calc_fib_recursive(n - 2)


@timer
def fib_recursive(n: int) -> int:
    return calc_fib_recursive(n)


fib_recursive(36)


@timer
def fib_loop(n: int) -> int:
    fib_1 = 1
    fib_2 = 2

    for _ in range(3, n + 1):
        fib_1, fib_2 = fib_2, fib_1 + fib_2

    return fib_2


fib_loop(36)


@timer
def fib_reduce(n: int) -> int:
    initial = (1, 0)
    return reduce(lambda acc, curr: (acc[0] + acc[1], acc[0]), range(n), initial)[0]


fib_reduce(36)
```

```python
def memoize(fn: Callable[[int], int]):
    cache: dict[int, int] = {}

    def inner(n: int) -> int:
        if n not in cache:
            cache[n] = fn(n)

        return cache[n]

    return inner


@memoize
def fib(n: int) -> int:
    print(f"Calculating fib({n})")
    return 1 if n < 3 else fib(n - 1) + fib(n - 2)


fib(10)
fib(10)
fib(35)


@memoize
def fact(n: int) -> int:
    print(f"Calculating factorial {n}")
    return 1 if n < 2 else n * fact(n - 1)


fact(5)
fact(5)
```

```python
from collections.abc import Callable


def timer(fn: Callable[..., int]):
    from functools import wraps
    from time import perf_counter

    @wraps(fn)
    def inner(*args: int, **kwargs: int):
        total_elapsed = 0
        result: int = 0

        for _ in range(10):
            start = perf_counter()
            result = fn(*args, **kwargs)
            end = perf_counter()
            total_elapsed += end - start

        avg_run_time = total_elapsed / 10

        args_ = ", ".join(map(str, args))
        kwargs_ = ",".join(f"{k}={v}" for k, v in kwargs.items())
        all_args = ", ".join(filter(None, [args_, kwargs_]))

        print(
            f"{fn.__name__}({all_args}) took average {avg_run_time:.6f} seconds ({10}) reps"
        )
        return result

    return inner


def calc_fib(n: int) -> int:
    return 1 if n < 3 else calc_fib(n - 1) + calc_fib(n - 2)


@timer
def fib(n: int) -> int:
    return calc_fib(n)


fib(28)
```

```python
def info(self: type):
    from datetime import datetime, timezone

    results: list[str] = []

    results.append(f"time: {datetime.now(timezone.utc)}")
    results.append(f"id: {hex(id(self))}")
    # results.append(f"class: {self.__class__.__name__}")
    results.append(f"class: {type(self).__name__}")
    for k, v in vars(self).items():
        results.append(f"{k}: {v}")

    return results


def debug_info(cls: type):
    cls.debug = info
    return cls


@debug_info
class Person:
    def __init__(self, name: str, birth_year: int) -> None:
        self.name = name
        self.birth_year = birth_year

    def say_hi(self):
        return "Hello there!"


p = Person("John", 1939)
p.debug()
```

```python
@debug_info
class Automobile:
    def __init__(self, make: str, model: str, year: int, top_speed: int) -> None:
        self.make = make
        self.model = model
        self.year = year
        self.top_speed = top_speed
        self._speed = 0

    @property
    def speed(self):
        return self._speed

    @speed.setter
    def speed(self, new_speed: int):
        if new_speed > self.top_speed:
            raise ValueError(f"Speed cannot exceed top speed of {self.top_speed}")
        else:
            self._speed = new_speed


favourite = Automobile("Tesla", "Model S", 2024, 200)
favourite.debug()
```

```python
from math import sqrt


def complete_ordering(cls: type):
    if "__eq__" in dir(cls) and "__lt__" in dir(cls):
        cls.__le__ = lambda self, other: self < other or self == other
        cls.__gt__ = lambda self, other: not (self < other) and not (self == other)
        cls.__ge__ = lambda self, other: not (self < other)

    return cls


@complete_ordering
class Point:
    def __init__(self, x: int, y: int) -> None:
        self.x = x
        self.y = y

    def __abs__(self) -> float:
        return sqrt(self.x**2 + self.y**2)

    def __repr__(self) -> str:
        return f"Point(x={self.x}, y={self.y})"

    def __eq__(self, other: object) -> bool:
        if isinstance(other, Point):
            return self.x == other.x and self.y == other.y
        return False

    def __lt__(self, other: object) -> bool:
        if isinstance(other, Point):
            return abs(self) < abs(other)
        return NotImplemented


p1, p2, p3, p4 = Point(2, 3), Point(2, 3), Point(0, 0), Point(100, 100)
abs(p1)
p1
p1 is p2
p1 == p2
p3 < p1
p4 > p1

p1 <= p4
p1 >= p3
p1 != p2
```
