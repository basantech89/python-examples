```python
def ginortS(s: str) -> str:
    return "".join(
        sorted(
            s,
            key=lambda c: (
                c.isdigit(),
                c.isdigit() and int(c) % 2 == 0,
                c.isupper(),
                c.islower(),
                c,
            ),
        )
    )


ginortS("Sorting1234")
```

```python
def find_mth_smallest(lst: list[list[int]], m: int):
    lists = [item for sublist in lst for item in sublist]
    lists.sort()
    return lists[m - 1]
```

```python
def is_prime(num: int) -> bool:
    for i in range(2, num):
        if num % i == 0:
            return False
    return True


is_prime(21)
```

```python
def goulds_sequence(n: int):
    return [bin(num).count("1") for num in range(1, n)]
```

```python
def first_unique_char(s: str):
    freqs: dict[str, int] = {}

    for c in s:
        freqs[c] = freqs.get(c, 0) + 1

    print(freqs)


first_unique_char("leetcode")
```

```python
def is_palindrome(s: str) -> bool:
    if len(s) == 1:
        return True

    if s[0] == s[len(s) - 1]:
        return is_palindrome(s[1:-1])

    return False


is_palindrome("racecar")
```
