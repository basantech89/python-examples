```python
import time
from collections.abc import Callable
from typing import Any


def time_it(
    fn: Callable[..., Any],
    *args: int,
    run_times: int = 1,
    **kwargs: int,
) -> float:
    start = time.perf_counter()

    for _ in range(run_times):
        fn(*args, **kwargs)

    end = time.perf_counter()
    return (end - start) / run_times


def compute_powers_1(n: int, *, start: int = 1, end: int) -> list[int]:
    results: list[int] = []

    for i in range(start, end):
        results.append(n**i)

    return results


def compute_powers_2(n: int, *, start: int = 1, end: int) -> list[int]:
    return [n**i for i in range(start, end)]


def compute_powers_3(n: int, *, start: int = 1, end: int) -> list[int]:
    return list(n**i for i in range(start, end))


# compute_powers_1(2, end=5)
# compute_powers_2(2, end=5)
# compute_powers_3(2, end=5)

time_it(compute_powers_1, 2, start=1, end=20000, run_times=5)
time_it(compute_powers_2, n=2, start=1, end=20000, run_times=5)
time_it(compute_powers_3, n=2, start=1, end=20000, run_times=5)
```

```python
def add_item(name: str, quantity: int, unit: str, grocery_list: list[Any] = []):
    grocery_list.append("{0} {1} {2}".format(name, quantity, unit))
    return grocery_list


store1 = add_item("banana", 2, "units")
add_item("milk", 1, "liter", store1)

store2 = add_item("python", 1, "medium-rare")
store1
store2
store1 is store2
```

```python
def add_item(
    name: str, quantity: int, unit: str, grocery_list: list[Any] | None = None
):
    if not grocery_list:
        grocery_list = []

    grocery_list.append("{0} {1} {2}".format(name, quantity, unit))
    return grocery_list


store1 = add_item("banana", 2, "units")
add_item("milk", 1, "liter", store1)

store2 = add_item("python", 1, "medium-rare")
store1
store2
store1 is store2
```

```python
def factorial(n: int, cache: dict[int, int] = {}) -> int:
    """compute the factorial of n using memoization

    inputs:
        n: int - the number to compute the factorial of
        cache: dict[int, int] - a dictionary to store previously computed factorials
    returns: int - the factorial of n
    """
    if n < 1:
        return 1

    if n in cache:
        return cache[n]

    print("calculating {0}!".format(n))
    result = n * factorial(n - 1)
    cache[n] = result
    return result


factorial(3)
factorial(3)
factorial(4)
```
