```python
import math


class FactIter:
    def __init__(self, n):
        self.n = n
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.i >= self.n:
            raise StopIteration

        result = math.factorial(self.i)
        self.i += 1
        return result


fact_iter = FactIter(5)
list(fact_iter)
list(fact_iter)
```

```python
def fact():
    i = 0

    def inner():
        nonlocal i
        result = math.factorial(i)
        i += 1
        return result

    return inner


f = fact()
fact_iter = iter(f, math.factorial(5))
list(fact_iter)
list(fact_iter)
```

```python
def fib(n):
    if n == 0:
        return 1

    fib_0 = 1

    result = 0
    prev_result = fib_0
    prev_prev_result = 0
    for _ in range(n):
        result = prev_result + prev_prev_result
        prev_prev_result = prev_result
        prev_result = result

    return result


fib(6)
[fib(i) for i in range(10)]
```

```python
class Squares:
    def __init__(self, n):
        self.n = n

    @staticmethod
    def squares(n):
        for i in range(n):
            yield i**2

    def __iter__(self):
        return Squares.squares(self.n)


sq = Squares(5)
for num in sq:
    print(num)

list(sq)
```

```python
from typing import NamedTuple


class Card(NamedTuple):
    rank: str
    suit: str


SUITS = ("Spaded", "Hearts", "Diamonds", "Clubs")
RANKS = tuple(range(2, 11)) + tuple("JQKA")


def card_gen():
    for i in range(len(SUITS) * len(RANKS)):
        suit = SUITS[i // len(RANKS)]
        rank = RANKS[i % len(RANKS)]
        yield Card(rank, suit)


def card_gen():
    for i in range(len(SUITS) * len(RANKS)):
        suit = SUITS[i // len(RANKS)]
        rank = RANKS[i % len(RANKS)]
        yield Card(rank, suit)


def card_gen():
    for suit in SUITS:
        for rank in RANKS:
            yield Card(rank, suit)


class CardDeck:
    SUITS = ("Spaded", "Hearts", "Diamonds", "Clubs")
    RANKS = tuple(range(2, 11)) + tuple("JQKA")

    def __iter__(self):
        return CardDeck.card_gen()

    @staticmethod
    def card_gen():
        for suit in CardDeck.SUITS:
            for rank in CardDeck.RANKS:
                yield Card(rank, suit)

    @staticmethod
    def reversed_card_gen():
        for suit in reversed(CardDeck.SUITS):
            for rank in reversed(CardDeck.RANKS):
                yield Card(rank, suit)

    def __reversed__(self):
        return CardDeck.reversed_card_gen()


deck = CardDeck()
list(reversed(deck))
```

```python
from math import factorial


def combo(n, k):
    return factorial(n) // (factorial(k) * factorial(n - k))


size = 10

pascal = [[combo(n, k) for k in range(n + 1)] for n in range(size + 1)]
pascal
```

```python
def matrix(n):
    return ((i * j for j in range(1, n + 1)) for i in range(1, n + 1))


m = list(matrix(5))


def matrix_iterator(n):
    for row in matrix(n):
        for item in row:
            yield item


for item in matrix_iterator(3):
    print(item)
```

```python
def matrix_iterator2(n):
    for row in matrix(n):
        yield from row


for item in matrix_iterator2(3):
    print(item)
```

```python
from datetime import date, datetime
from functools import partial
from typing import Any, NamedTuple

file_name = "/Users/basantsoni/Downloads/nyc_parking_tickets_extract.csv"
with open(file_name) as f:
    columns = next(f).strip("\n").split(",")
    headers = [column.replace(" ", "_").lower() for column in columns]
    sample_data = next(f).strip("\n").split(",")


def get_parking_fields(headers: list[str]):
    fields: list[tuple[str, Any]] = []

    for header in headers:
        header = header.replace(" ", "")
        if header == "issue_date":
            fields.append((header, date))
        elif header == "violation_code" or header == "summons_number":
            fields.append((header, int))
        else:
            fields.append((header, str))

    return fields


class ParkingTicket(NamedTuple):
    summons_number: int
    plate_id: str
    registration_state: str
    plate_type: str
    issue_date: date
    violation_code: int
    vehicle_body_type: str
    vehicle_make: str
    violation_description: str


def parse_int(value: Any, *, default: Any = None):
    try:
        return int(value)
    except ValueError:
        return default


def parse_date(value: Any, *, default: Any = None):
    date_format = "%m/%d/%Y"
    try:
        return datetime.strptime(value, date_format).date()
    except ValueError:
        return default


def parse_string(value: Any, *, default: Any = None):
    try:
        cleaned = value.strip()
        if not cleaned:
            return default

        return cleaned
    except ValueError:
        return default


column_parsers = (
    parse_int,
    parse_string,
    # lambda s: parse_string(s, default=""),
    partial(parse_string, default=""),
    partial(parse_string, default=""),
    parse_date,
    parse_int,
    partial(parse_string, default=""),
    parse_string,
    partial(parse_string, default=""),
)


def parse_row(row: str, *, default: Any = None):
    fields = row.strip("\n").split(",")
    parsed_data = [func(field) for func, field in zip(column_parsers, fields)]

    if all(item is not None for item in parsed_data):
        return ParkingTicket(*parsed_data)

    return default


def read_data():
    with open(file_name) as f:
        next(f)
        yield from f


def parse_data():
    for row in read_data():
        parsed = parse_row(row)
        if parsed:
            yield parsed


parsed_rows = parse_data()


def get_violations_by_car_make():
    parsed_rows = parse_data()
    violations = {}

    for row in parsed_rows:
        make = row.vehicle_make
        violations[make] = violations.get(make, 0) + 1

    return dict(sorted(violations.items(), key=lambda x: x[1], reverse=True))


get_violations_by_car_make()
```
