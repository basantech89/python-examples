```python
import ctypes
import gc
```

```python
def ref_count(address: int) -> int:
    """Return the reference count of an object."""
    return ctypes.c_long.from_address(address).value
```

```python
def object_by_id(object_id: int) -> str:
    for obj in gc.get_objects():
        if id(obj) == object_id:
            return "Object exists"
    return "Not Found"
```

```python
class A:
    def __init__(self) -> None:
        self.b = B(self)
        print(f"A: self: {hex(id(self))}, b: {hex(id(self.b))}")
```

```python
class B:
    def __init__(self, a: "A") -> None:
        self.a = a
        print(f"B: self: {hex(id(self))}, a: {hex(id(self.a))}")
```

```python
# Disable automatic garbage collection so we can see what's going on,
# bcoz if we don't disable it, then as references go away, it's gonna clean up the circular references
gc.disable()
```

```python
my_var = A()
```

```python
print(hex(id(my_var.b)))
print(hex(id(my_var.b.a)))
print(hex(id(my_var)))
```

```python
a_id = id(my_var)
b_id = id(my_var.b)
```

```python
ref_count(a_id)
```

```python
ref_count(a_id), ref_count(b_id)
```

```python
object_by_id(a_id)
```

```python
object_by_id(b_id)
```

```python
my_var = None
```

```python
ref_count(a_id)
```

```python
ref_count(b_id)
```

```python
object_by_id(a_id), object_by_id(b_id)
```

```python
object_by_id(a_id), ref_count(a_id), gc.is_tracked(a_id)
```
