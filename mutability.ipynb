```python
def process(s: str) -> None:
    print(f"initial id(s): {id(s)}")
    s += " world"
    print(f"final id(s): {id(s)}")


my_var = "hello"
print(f"initial id(my_var): {id(my_var)}")
process(my_var)
print(f"final id(my_var): {id(my_var)}")
print(f"my_var: {my_var}")
```

```python
import builtins

LST = builtins.list[int]


def process_list(lst: LST) -> None:
    print(f"initial id(lst): {id(lst)}")
    lst.append(4)
    print(f"final id(lst): {id(lst)}")


my_list = [1, 2, 3]
print(f"initial id(my_list): {id(my_list)}")
process_list(my_list)
print(f"final id(my_list): {id(my_list)}")
```

```python
import builtins

TUPLE = builtins.tuple[list, str]


def process_tuple(t: TUPLE) -> None:
    print(f"initial id(t): {id(t)}")
    t[0].append(5)
    print(f"final id(t): {id(t)}")


my_tuple = ([1, 2], "hello")
print(f"initial id(my_tuple): {id(my_tuple)}")
process_tuple(my_tuple)
print(f"final id(my_tuple): {id(my_tuple)}")
```

```python
import ctypes
import gc


def ref_count(address: int) -> int:
    """Return the reference count of an object."""
    return ctypes.c_long.from_address(address).value


def object_by_id(object_id: int) -> str:
    for obj in gc.get_objects():
        if id(obj) == object_id:
            return "Object exists"
    return "Not Found"


a1 = "hello"
b1 = "hello"

print(id(a1))
print(id(b1))
print(ref_count(id(a1)))
print(ref_count(id(b1)))
print(object_by_id(id(a1)))
print(object_by_id(id(b1)))
```
